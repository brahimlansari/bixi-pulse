```python
BRONZE = "abfss://bronze@stbixipulse2611.dfs.core.windows.net"
```

```python
trips = spark.read.parquet(f"{BRONZE}/bixi_trips")
print(f"{trips.count():,} trajets")
trips.printSchema()
```

```python
display(trips.limit(5))
```

```python
from pyspark.sql import functions as F

status = spark.read.parquet(f"{BRONZE}/station_status")

display(
    status.groupBy("ingestion_date")
    .agg(
        F.countDistinct("ingested_at").alias("nb_snapshots"),
        F.count("*").alias("nb_lignes"),
    )
    .orderBy("ingestion_date")
)
```

```python
weather = spark.read.parquet(f"{BRONZE}/weather_hourly")
print(f"Lignes totales    : {weather.count():,}")
print(f"Heures distinctes : {weather.select('observed_at').distinct().count():,}")
```
